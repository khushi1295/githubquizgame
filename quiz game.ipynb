```python
# 🎯 QUIZ GAME PROJECT by Khushboo Verma

def run_quiz(questions):
    score = 0
    print("Welcome to the Python Quiz Game!\n")
    for question, options, answer in questions:
        print(question)
        for option in options:
            print(option)
        user_answer = input("Enter your answer (A/B/C/D): ").strip().upper()

        if user_answer == answer:
            print("✅ Correct!\n")
            score += 1
        else:
            print(f"❌ Wrong! The correct answer was {answer}\n")

    print(f"🎉 You got {score} out of {len(questions)} correct.")
    print("Thanks for playing!\n")


# Question Bank: (question, options, correct_answer)
quiz_questions = [
    ("1. What is the capital of India?",
     ["A. Mumbai", "B. New Delhi", "C. Kolkata", "D. Chennai"], "B"),

    ("2. Which language is this quiz written in?",
     ["A. Java", "B. C++", "C. Python", "D. HTML"], "C"),

    ("3. Who developed Python?",
     ["A. Elon Musk", "B. Guido van Rossum", "C. Dennis Ritchie", "D. Bill Gates"], "B"),

    ("4. What does CPU stand for?",
     ["A. Central Process Unit", "B. Central Processing Unit", "C. Computer Personal Unit", "D. None"], "B"),

    ("5. Which keyword is used to define a function in Python?",
     ["A. def", "B. func", "C. define", "D. function"], "A")
]


# Run the Quiz
while True:
    run_quiz(quiz_questions)
    play_again = input("Do you want to play again? (yes/no): ").strip().lower()
    if play_again != "yes":
        print("👋 Goodbye! See you next time.")
        break
```
